```python
import pandas as pd
import matplotlib.pyplot as plt

# Function to process the data
def process_data(file_name):
    # Load the CSV file
    data = pd.read_csv(file_name + '.csv')

    # Ensure 'RT' and 'm/z' columns are numeric, coercing errors to NaN
    data['RT'] = pd.to_numeric(data['RT'], errors='coerce')
    data['m/z'] = pd.to_numeric(data['m/z'], errors='coerce')

    # Filter out rows with NaN values in 'RT' or 'm/z' columns
    cleaned_data = data.dropna(subset=['RT', 'm/z'])

    # Filter the cleaned data to consider only RT values until 30.000
    cleaned_data = cleaned_data[(cleaned_data['RT'] <= 38.000) & (cleaned_data['RT'] > 5.000)]

    # Normalize 'm/z' values to a 1-100 scale
    cleaned_data['m/z'] = ((cleaned_data['m/z'] - cleaned_data['m/z'].min()) / 
                            (cleaned_data['m/z'].max() - cleaned_data['m/z'].min())) * 100 + 1
    return cleaned_data

# File names and corresponding colors
files_and_colors = [
    ('S1MS(0.5)', 'red'),
    ('S2MS(0.4)', 'blue'),
    ('S3MS(0.3)', 'green'),
    ('S5MS(0.1)', 'orange'),
    ('S4GC', 'purple')
]

# Create a plot with increased width and decreased height
plt.figure(figsize=(15, 5))

# Loop through each file and plot
for file_name, color in files_and_colors:
    data = process_data(file_name)
    plt.plot(data['RT'], data['m/z'], color=color, linewidth=0.5, label=file_name)

# Customize the plot
plt.xlabel('Time (min)')
plt.ylabel('Relative Abundance')
plt.title('Overlapped GCMS Spectral Graphs')
plt.legend()

# Save the plot as a PNG file with 300 dpi
plt.savefig('Overlapped_GCMS_Spectral_Graphs.png', dpi=300)

# Display the plot
plt.show()
```
