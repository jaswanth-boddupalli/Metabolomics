```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import find_peaks

# Read the CSV file
data = pd.read_csv("JA_C_N.csv")


# Filter data based on RT values between 5.0 and 30.0 and m/z values below 5436282
data = data[(data['RT'] <= 30.000) & (data['RT'] > 5.000)]
data = data[data['m/z'] <= 5436282]

# Normalize 'm/z' values to a 1-100 scale
data['m/z'] = ((data['m/z'] - data['m/z'].min()) / (data['m/z'].max() - data['m/z'].min())) * 100 + 1

# Separate the RT and m/z values
x = data['RT']
y = data['m/z']

# Find peaks in the data using scipy's find_peaks function
# The prominence parameter helps control how "significant" a peak must be to be labeled
peaks, _ = find_peaks(y, prominence=3)

# Create a plot
plt.figure(figsize=(15, 12))  # Adjusting the plot size to match your requirement

# Plot the RT vs m/z
plt.plot(x, y, 'k-', linewidth=0.5)

# Add labels for the peaks
for peak in peaks:
    plt.text(x.iloc[peak], y.iloc[peak] + 5, f'{x.iloc[peak]:.2f}', ha='center')

# Customize the plot
plt.xlabel('Time (min)')
plt.ylabel('Relative Abundance')
plt.title('JA_Control GCMS Spectral Graph')

# Save the plot
plt.savefig('JA_Control_GCMS_Spectral_Graph.png', dpi=300)

# Show the plot
plt.show()
```
